```python
import pandas as pd
import numpy as np

# Muat Dataset Titanic
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

```

```python
# 1 Berapa jumlah baris dan kolom dataset Titanic?
print(df.shape, df.columns.tolist())

```

```python
# 2 Kolom mana yang memiliki nilai kosong ?

print(df.isnull().sum())
```

```python
# 3 Berapa persen penumpang yang selamat (Survived=1)?
print(df['Survived'].mean() * 100)
```

```python
# 4 Rata-rata usia penumpang wanita kelas 1 yang selamat?
#filter
penumpang_wanita_kelas1_selamat = df[(df["Sex"] == 'female') & (df['Pclass'] == 1) & (df['Survived'] == 1)]

#rata-rata usianya
rata_rata_usia = penumpang_wanita_kelas1_selamat['Age'] .mean()
print(f'Rata-rata usia penumpang wanita kelas 1 yang selamat adalah : {rata_rata_usia:.2f} tahun')
```

```python
# 5 Kelas penumpang mana yang tingkat keselamatan tertinggi?
tingkat_keselamatan_per_kelas = df.groupby('Pclass')['Survived'].mean() .sort_values(ascending = False)
print(tingkat_keselamatan_per_kelas)
```

```python
# 6 jumlah orang yang selamat dan tidak selamat berdasarkan kelas
jumlah_selamat_dan_tidak_selamat_per_kelas = df.groupby(['Pclass', 'Survived']).size().unstack(fill_value=0)

jumlah_selamat_dan_tidak_selamat_per_kelas.columns = ['Tidak Selamat', 'Selamat']
print(jumlah_selamat_dan_tidak_selamat_per_kelas)
```
