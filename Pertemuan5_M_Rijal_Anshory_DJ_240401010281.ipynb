```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, matplotlib.gridspec as gridspec
import seaborn as sns
# 1 load data set
df = sns.load_dataset('titanic')
print (df.shape, df.dtypes)
print (df.describe().round(2))

# 2 setup figure layout
fig = plt.figure(figsize=(16, 12))
fig.suptitle(
    'Dashboard Analisis Dataset Titanic', fontsize=18, fontweight='bold', y=0.98
)

gs = gridspec.GridSpec(2, 2, figure=fig, hspace=0.4, wspace=0.35)
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, 0])
ax4 = fig.add_subplot(gs[1, 1])

# 3 BAR CHART
sns.countplot(
    data=df, x='class', palette='Set2', ax=ax1
)

ax1.set_title('Jumlah Penumpang Berdasarkan Kelas')
ax1.set_xlabel('Class')
ax1.set_ylabel('Jumlah Penumpang')

# 4 HISTOGRAM
sns.histplot(
    data=df, x='age', bins=20, kde=True, color='skyblue', ax=ax2
)

ax2.set_title('Distribusi Umur Penumpang')
ax2.set_xlabel('umur')
ax2.set_ylabel('Jumlah Penumpang')

# 5 BOXPLOT
sns.boxplot(
    data=df, x='class', y='age', palette='Set3', ax=ax3
)

ax3.set_title('Distribusi Usia Penumpang Berdasarkan Kelas')
ax3.set_xlabel('Class')
ax3.set_ylabel('Usia')

# 6 SCATTER
sns.scatterplot(
    data=df, x='age', y='fare', hue='survived', palette='Set1', ax=ax4
)

ax4.set_title('Hubungan Usia dan Tarif Tiket')
ax4.set_xlabel('Usia')
ax4.set_ylabel('fare (USD)')
ax4.legend(title='Survived')

# 7 Tampilkan Dashboard
plt.tight_layout()
plt.show()

# 8 Ekspor
plt.savefig('dashboard_titanic_byRijal.png', dpi=150, bbox_inches='tight')
plt.show()
print('Dashboard berhasil di export')
```
